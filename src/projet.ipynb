Propagation d'une rumeur - Célia, Ivan, Karene

Il peut être intéressant d'étudier la propagation d'une rumeur au sein d'un lycée parisien et d'observer comment certains facteurs influencent cette propagation. 

Ce notebook présente ici le modèle de propagation d'une rumeur sous différentes parties: 

    1. Modèle simple: Cette première partie est une sorte d'introduction au modèle de manière aléatoire et simple car il n'y a pas de paramètres complexes.

    2. Modèle complexe incluant le seuil d'amitié: Cette deuxième partie est beaucoup plus poussée que la première car, ici, le seuil d'amitié rentre en jeu.

    3. Modèle encore plus complexe incluant les oubliants: Cette troisième partie est encore plus complexe que la précédente car on y ajoute un nouveau paramètre qui est l'oubliant. 

Tout ces modèles sont intéressants dans notre analyse car, à leurs échelles, permettent de distinguer comment des paramètres influencent un système dynamique. 

I- Modèle Simple- Propagation d'une rumeur aléatoirement

```python
import numpy as np
import random

from matplotlib import pyplot as plt
```

```python
#Paramètres du modèle de propagation d'une rumeur
#Paramètres initiaux, changeront par la suite
size=475 #(1)
lycee=[]
lycee_ami = []
createur=1 #la personne créant la rumeur
oubliant=0 #la rumeur n'a pas débuté
propagateur=0 #la rumeur n'a pas débuté
ignorant=474 #presque tout le monde, car la rumeur n'a pas débuté
max_tours=60 #correspond à 1 mois
jours=0
prob_oubli = 0.1  
```

```python
def world_lycee(lycee,size):
    lycee.append("C")
    for i in range(size-1):
        lycee.append("I")
    return lycee

lycee = world_lycee(lycee,size)
```

```python
def get_propagateurs(lycee):
    propagateurs=[]
    for i in range(len(lycee)):
        if lycee[i]=='P' or lycee[i]=='C':
            propagateurs.append(i)
    return propagateurs
print(get_propagateurs(lycee))
```

```python
def get_ignorants(lycee):
    ignorants=[]
    for i in range(len(lycee)):
        if lycee[i]=='I':
            ignorants.append(i)
    return ignorants
print(get_ignorants(lycee))
```

```python
def propagation(lycee):
    propagateurs = get_propagateurs(lycee)
    for p in propagateurs:
        ignorants = get_ignorants(lycee)
        if len(ignorants) > 0:  
            personnes_rencontrees=np.random.choice(ignorants,random.randint(0,30))
            for j in personnes_rencontrees:
                lycee[j] = "P"
    return lycee
print (propagation(lycee))
```

```python
def nb_jours(lycee):
    jours = 0
    while "I" in lycee:
        lycee = propagation(lycee)
        jours += 1
        if jours > max_tours:  
            return max_tours
    return jours
print((nb_jours(lycee)))
```

```python
#Ceci est le graphique de la fonction précédente (propagation). 
#Cela permet d'avoir un visuel beaucoup plus clair, parlant et compréhensible pour ce modèle simple.
# On remarque que le nombre d'ignorants chute rapidemtn car le propagateur peut rencontrer jusqu'à 30 personnes en seulement 1 jour.
#On a recrée un lycée neuf, afin d'avoir une simulation démarrant de zéro.
# On a également ajouté deux listes vistes pour les ignorants et propagateurs afin de stocker l'évolution au fil des jours.
lycee = world_lycee([], size)  
liste_ignorants = []
liste_propagateurs = []

for i in range(max_tours):
    liste_ignorants.append(lycee.count("I"))
    liste_propagateurs.append(lycee.count("P") + lycee.count("C"))
    lycee = propagation(lycee)

plt.plot(liste_ignorants, label="Ignorants", color="blue")
plt.plot(liste_propagateurs, label="Propagateurs", color="green")
plt.xlabel("Nombre de jours")
plt.ylabel("Nombre de personnes")
plt.title("Propagation de la rumeur")
plt.legend()
plt.grid(True)
plt.show()
```

Partie II- Modèle complexe incluant le seuil d'amitié

```python
lycee_niv_ami = []
lycee_ami = []
seuil= 0.9
```

```python
#Ici on a créé une fonction pour crée une liste avec des chiffres compris entre 0 et 0.9.Cette liste correspond au niveau d'amabilité d'une personne.
def lycee_avec_amis(size):
    lycee = []
    lycee.append(0.0)
    for i in range(size-1):
        eleve = round(random.random(),2)
        while eleve == 0.0:
            eleve = round(random.random(),2)
        lycee.append(eleve)
    return lycee
```

```python
lycee_niv_ami = lycee_avec_amis(size)
lycee_ami = world_lycee(lycee_ami,size)
```

On introduit la notion de seuil. On fait correspondre chaque valeur au niveau d'amitié qu'une personne au hazard à avec le créateur.

```python
#C'est le même principe qu'avec la fonction propagateur sauf qu'ici une personne est propagateur si la somme des niveaux d'amabilités des personnes
# est supérieur ou égale au seuil.
def propagation_avec_ami(lycee_ami, lycee_niv_ami, seuil):
    propagateurs = get_propagateurs(lycee_ami)
    for p in propagateurs:
        ignorants = get_ignorants(lycee_ami)
        if len(ignorants) > 0: 
            personnes_rencontrees = np.random.choice(ignorants, min(len(ignorants), random.randint(0, 30)), replace=False)
            for j in personnes_rencontrees:
                if lycee_niv_ami[p] + lycee_niv_ami[j] >= seuil:
                    lycee_ami[j] = "P"
    return lycee_ami
```

```python
def nb_jours_ami(lycee_ami,lycee_niv_ami,seuil):
    jours = 0
    while "I" in lycee_ami:  
        lycee_ami = propagation_avec_ami(lycee_ami,lycee_niv_ami,seuil)
        jours += 1 
        if jours > max_tours:  
            return max_tours
    return jours
```

```python
lycee_n = world_lycee([], size) 
lycee_niv_ami_n = lycee_avec_amis(size)
```

```python
def simulation_avec_ami(lycee_n,lycee_niv_ami_n,size, seuil, max_tours):
    stats = {"propagateurs": [], "ignorants": []}
    jours = 0
    while "I" in lycee_n and jours < max_tours: 
        propagateurs = get_propagateurs(lycee_n)
        ignorants = get_ignorants(lycee_n)

        stats["propagateurs"].append(len(propagateurs))
        stats["ignorants"].append(len(ignorants))

        lycee_n = propagation_avec_ami(lycee_n, lycee_niv_ami_n, seuil)
        jours += 1
        
    propagateurs = get_propagateurs(lycee_n)
    ignorants = get_ignorants(lycee_n)
    stats["propagateurs"].append(len(propagateurs))
    stats["ignorants"].append(len(ignorants))
    return stats


stats = simulation_avec_ami(lycee_n,lycee_niv_ami_n,size, seuil, max_tours)

plt.figure(figsize=(10, 6))
plt.plot(stats["propagateurs"], label="Propagateurs", color="blue")
plt.plot(stats["ignorants"], label="Ignorants", color="red")
plt.title("Évolution de la propagation de la rumeur avec amitié")
plt.xlabel("Jours")
plt.ylabel("Nombre d'individus")
plt.legend()
plt.grid()
plt.show()
```

```python
def generer_liste_seuil():
    return list(np.arange(0.1, 2.0, 0.1))
```

```python
lycee_n = world_lycee([], size) 
lycee_niv_ami_n = lycee_avec_amis(size)
```

```python
def differents_seuil(lycee_n,lycee_niv_ami_n):
    seuils= generer_liste_seuil()
    Jours = []
    for seuil in seuils:
        lycee_n = world_lycee([], size) 
        lycee_niv_ami_n = lycee_avec_amis(size)
        nb_jours = nb_jours_ami(lycee_n,lycee_niv_ami_n,seuil)
        Jours.append(nb_jours)
    return Jours, seuils
        


jours, seuils = differents_seuil(lycee_n, lycee_niv_ami_n)

# Création du graphique
plt.plot(seuils, jours, marker='o')  # X: jours, Y: seuils
plt.xlabel("Seuil")
plt.ylabel("Nombre de jours")
plt.title("Évolution du seuil en fonction du nombre de jours")
plt.grid(True)
plt.show()
```

```python
def simulations_propagation(nb_simulations, size, seuil):
    #La fonction permet d'observer la variabilité de la propagation total d'une rumeur sur plusieurs simulations effectués
    durees = []
    for _ in range(nb_simulations):
        lycee_ami = world_lycee([], size)
        lycee_niv_ami = lycee_avec_amis(size)
        durees.append(nb_jours_ami(lycee_ami, lycee_niv_ami, seuil))
    return durees
    
print(simulations_propagation(3, 495, 1.0))
```

```python
#Histogramme de la fonction précédente
#Cet histogramme permet de visualiser les différentes simulations effectués.
#Cela permet d'observer que lors de chaque simulation, la rumeur se propage sur un nombre de jours différents (=influence et intérets des paramètres)
durees = simulations_propagation(3, 495, 1.0) # remplace par les résultats de ta simulation

plt.hist(durees, bins=range(min(durees), max(durees) + 2), edgecolor='black') 
plt.title("Distribution des durées de propagation")
plt.xlabel("Nombre de jours")
plt.ylabel("Fréquence")
plt.grid(True)
plt.show()
```

```python
def contacts_necessaires(size, seuil, essais):
#La fonction permet de connaitre le nombre de contacts qu'il faut en moyenne pour qu'un élève ignorant devienne propagateur
    essais=5
    total_contacts = 0
    total_propagateurs = 0

    for _ in range(essais):
        lycee_ami = world_lycee([], size)
        lycee_niv_ami = lycee_avec_amis(size)

        jours = 0
        while "I" in lycee_ami and jours < max_tours:
            propagateurs = get_propagateurs(lycee_ami)
            ignorants = get_ignorants(lycee_ami)
            for p in propagateurs:
                personnes_rencontrees = np.random.choice(ignorants, random.randint(0, 30), replace=True)
                total_contacts += len(personnes_rencontrees)
            total_propagateurs += len(propagateurs)

            lycee_ami = propagation_avec_ami(lycee_ami, lycee_niv_ami, seuil)
            jours += 1

    if total_propagateurs == 0:
        return 0  
    return round(total_contacts / total_propagateurs, 2)
    
print(contacts_necessaires(size, 1.0,5))
```

```python
#Le graphique utilise la fonction précédente afin d'avoir un visuel beaucoup plus clair 
#Il montre l'impact du seuil d'amitié sur la difficulté à propager une rumeur
seuils = [0.1 * i for i in range(1, 11)]
contacts = [contacts_necessaires(size, s, essais=3) for s in seuils]

plt.plot(seuils, contacts, marker='o', color='purple')
plt.title("Nombre moyen de contacts nécessaires selon le seuil")
plt.xlabel("Seuil d'amitié")
plt.ylabel("Contacts moyens par propagateur")
plt.grid(True)
plt.show()
```

Partie III- Modèle très complexe incluant les oubliants

```python
#Paramètres du modèle de propagation d'une rumeur
#Paramètres initiaux, changeront par la suite
size=475 #(1)
lycee_ami2 = []
lycee2= []
createur=1 #la personne créant la rumeur
oubliant=0 #la rumeur n'a pas débuté
propagateur=0 #la rumeur n'a pas débuté
ignorant=474 #presque tout le monde, car la rumeur n'a pas débuté
max_tours=60 #correspond à 1 mois
jours=0
prob_oubli = 0.1  
seuil= 0.9

```

```python
def propagation_avec_ami_et_oubliants(lycee, lycee_ami, seuil, prob_oubli):
    propagateurs = get_propagateurs(lycee)
    for p in propagateurs:
        ignorants = get_ignorants(lycee)
        if len(ignorants) > 0:
            personnes_rencontrees = np.random.choice(ignorants, min(len(ignorants), random.randint(1, 30)), replace=False)
            for j in personnes_rencontrees:
                if (random.random() < prob_oubli):
                    lycee[p] = "O"  # Le propagateur oublie
                elif (lycee_ami[p] + lycee_ami[j] >= seuil) and (lycee[j] != "O"):
                    lycee[j] = "P"
    return lycee
propagation_avec_ami_et_oubliants(lycee2, lycee_ami2, seuil, prob_oubli)   
```

```python

def nb_jours_ami_et_oubliants(seuil, prob_oubli, max_tours):
    jours = 0
    lycee_ami= lycee_avec_amis(size)
    lycee= world_lycee([],size)
    while "I" in lycee:
        lycee = propagation_avec_ami_et_oubliants(lycee, lycee_ami, seuil, prob_oubli)
        jours += 1
        if jours > max_tours:
            return max_tours
    return jours
print(nb_jours_ami_et_oubliants(seuil, prob_oubli, max_tours))
```

```python
def simulation_avec_oubliants(size, seuil, prob_oubli, max_tours):
    lycee = world_lycee([], size) 
    lycee_ami = [0.0] + [round(random.random(), 2) for _ in range(size - 1)]
    stats = {"propagateurs": [], "ignorants": [], "oublieurs": []}

    for jour in range(max_tours):
        stats["propagateurs"].append(lycee.count("P"))
        stats["ignorants"].append(lycee.count("I"))
        stats["oublieurs"].append(lycee.count("O"))
        if "I" not in lycee: 
            break
        lycee = propagation_avec_ami_et_oubliants(lycee, lycee_ami, seuil, prob_oubli)
    
    return stats

stats = simulation_avec_oubliants(size, seuil, prob_oubli, max_tours)
plt.figure(figsize=(10, 6))
plt.plot(stats["propagateurs"], label="Propagateurs", color="blue")
plt.plot(stats["ignorants"], label="Ignorants", color="red")
plt.plot(stats["oublieurs"], label="Oublieurs", color="green")
plt.title("Évolution de la propagation avec oubliants")
plt.xlabel("Jours")
plt.ylabel("Nombre d'individus")
plt.legend()
plt.grid()
plt.show()
```
